## 1. Load and Explore the Dataset

**What we do here**
- Import the libraries (pandas, numpy, matplotlib).
- Load the CSV. The file is a PDF-style export: the real header is spread over 6 lines, so we skip them (`skiprows=6`) and give our own column names. The last 4 columns (`c17` to `c20`) are temporary names, we fix them in the cleaning step.
- Look at the first and last rows, the shape, data types and missing values.

```python
import glob
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---- file path: put your csv in the same folder as this notebook ----
CSV_PATH = "TATA IPL 2025- Auction List -15.11.24.csv"
if not os.path.exists(CSV_PATH):          # if the name is different, find it automatically
    found = [f for f in glob.glob("*.csv") if "ipl" in f.lower() and "clean" not in f.lower()]
    if found:
        CSV_PATH = found[0]
print("Using file:", CSV_PATH)

# ---- load (skip the 6 header lines, give our own column names) ----
cols = ["Sr_No", "Set_No", "Set_2025", "First_Name", "Surname", "Country",
        "State_Assoc", "DOB", "Age", "Specialism", "Batting", "Bowling",
        "Test_Caps", "ODI_Caps", "T20_Caps", "IPL_Matches", "Prev_IPL_Teams",
        "c17", "c18", "c19", "c20"]
raw = pd.read_csv(CSV_PATH, skiprows=6, header=None, names=cols, dtype=str)

# ---- explore ----
print("\nShape (rows, columns):", raw.shape)

print("\nFirst 5 rows:")
display(raw.head())

print("Last 5 rows (the last row is a page number, not a player):")
display(raw.tail())

print("Data types and non-null counts:")
raw.info()

print("\nMissing values per column:")
display(raw.isna().sum().to_frame("missing"))
```

---
## 2. Clean the Dataset

**Problems found in the raw data and how we fix them**
1. **Page-break rows** (titles, repeated headers, page numbers) -> keep only rows with a numeric `Sr_No`.
2. **Extra spaces** in text -> strip them.
3. **Shifted columns** - the last 4 fields move left when a value is blank. We find the `Capped / Uncapped / Associate` word and read the base price from the column after it.
4. **Wrong data types** (numbers stored as text, DOB as text) -> convert.
5. **Missing values** - blank caps / IPL matches mean 0, other blanks get a clear label.
6. **Inconsistent text** - `RCB.KKR` -> `RCB,KKR`, `Mi` -> `MI`, same capital letters for roles.
7. **Duplicates** -> check and remove.
8. **New columns** - full name, Indian/Overseas, total international caps, number of previous IPL teams.

```python
# 1. remove page-break rows (real players always have a number in Sr_No)
df = raw[raw["Sr_No"].fillna("").str.strip().str.isdigit()].copy().reset_index(drop=True)
print("Rows removed:", len(raw) - len(df), "| Players left:", len(df))

# 2. remove extra spaces
df = df.apply(lambda col: col.str.strip())

# 3. fix the shifted columns
def fix_row(r):
    for i, c in enumerate(["c17", "c18", "c19"]):
        if r[c] in ("Capped", "Uncapped", "Associate"):
            return pd.Series({
                "Cap_Status": r[c],
                "Base_Price_Lakh": r[["c18", "c19", "c20"][i]],
                "Team_2024": r["c17"] if i >= 1 else np.nan,
                "IPL_2024_Matches": r["c18"] if i == 2 else np.nan})
    return pd.Series({"Cap_Status": np.nan, "Base_Price_Lakh": np.nan,
                      "Team_2024": np.nan, "IPL_2024_Matches": np.nan})

fixed = df.apply(fix_row, axis=1)
df = pd.concat([df.drop(columns=["c17", "c18", "c19", "c20"]), fixed], axis=1)

# 4. fix data types
num_cols = ["Sr_No", "Set_No", "Age", "Test_Caps", "ODI_Caps", "T20_Caps",
            "IPL_Matches", "IPL_2024_Matches", "Base_Price_Lakh"]
for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df["DOB"] = pd.to_datetime(df["DOB"], format="%d/%m/%Y", errors="coerce")

# 5. missing values
count_cols = ["Test_Caps", "ODI_Caps", "T20_Caps", "IPL_Matches", "IPL_2024_Matches"]
df[count_cols] = df[count_cols].fillna(0).astype(int)      # blank = 0
df["State_Assoc"] = df["State_Assoc"].fillna("Overseas / N.A.")
df["Batting"] = df["Batting"].fillna("Not specified")
df["Bowling"] = df["Bowling"].fillna("Not specified")
df["Team_2024"] = df["Team_2024"].fillna("None")
df["Prev_IPL_Teams"] = df["Prev_IPL_Teams"].fillna("")

# 6. make text consistent
df["Prev_IPL_Teams"] = (df["Prev_IPL_Teams"].str.replace(".", ",", regex=False)
                        .str.replace(" ", "", regex=False).str.upper())
df["Team_2024"] = df["Team_2024"].replace({"Mi": "MI"})
df["Specialism"] = df["Specialism"].str.title().replace({"All-Rounder": "All-rounder"})
df["Bowling"] = df["Bowling"].str.title()

# 7. duplicates
print("Duplicate players:", df.duplicated(["First_Name", "Surname", "DOB"]).sum())
df = df.drop_duplicates(["First_Name", "Surname", "DOB"]).reset_index(drop=True)

# 8. new columns
df["Player"] = df["First_Name"] + " " + df["Surname"]
df["Nationality"] = np.where(df["Country"] == "India", "Indian", "Overseas")
df["Total_Intl_Caps"] = df["Test_Caps"] + df["ODI_Caps"] + df["T20_Caps"]
df["Prev_Teams_Count"] = df["Prev_IPL_Teams"].apply(lambda s: len([t for t in s.split(",") if t]))

# check and save
print("Cleaned shape:", df.shape)
print("Missing values left:", int(df.drop(columns=["DOB"]).isna().sum().sum()))
df.to_csv("ipl_2025_auction_cleaned.csv", index=False)
print("Saved: ipl_2025_auction_cleaned.csv")

print("\nCleaned data (first 5 rows):")
display(df.head())

print("Summary statistics:")
display(df[["Age", "Test_Caps", "ODI_Caps", "T20_Caps", "IPL_Matches", "Base_Price_Lakh"]].describe().round(1))

print("Players by role / cap status / nationality:")
display(pd.concat([df["Specialism"].value_counts(), df["Cap_Status"].value_counts(),
                   df["Nationality"].value_counts()], axis=1, keys=["Role", "Cap status", "Nationality"]).fillna(""))
```

## 3. Data Visualization

Six simple charts:
1. Top 10 countries
2. Players by role
3. Base price tiers
4. Age distribution
5. Capped vs uncapped players by role
6. Correlation heatmap of the numeric columns

```python
BLUE, ORANGE = "#2a78d6", "#eb6834"

# 1. players by country (top 10)
plt.figure(figsize=(8, 4.5))
df["Country"].value_counts().head(10).sort_values().plot(kind="barh", color=BLUE)
plt.title("Top 10 countries in the auction pool")
plt.xlabel("Number of players")
plt.tight_layout()
plt.show()

# 2. players by role
plt.figure(figsize=(7, 4.5))
ax = df["Specialism"].value_counts().plot(kind="bar", color=BLUE)
ax.bar_label(ax.containers[0])
plt.title("Players by role")
plt.ylabel("Number of players")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 3. base price tiers
plt.figure(figsize=(8, 4.5))
ax = df["Base_Price_Lakh"].astype(int).value_counts().sort_index().plot(kind="bar", color=BLUE)
ax.bar_label(ax.containers[0])
plt.title("Base price tiers (Rs lakh)")
plt.xlabel("Base price (Rs lakh)")
plt.ylabel("Number of players")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 4. age distribution
plt.figure(figsize=(8, 4.5))
plt.hist(df["Age"], bins=range(int(df["Age"].min()), int(df["Age"].max()) + 2),
         color=BLUE, edgecolor="white")
plt.axvline(df["Age"].median(), color=ORANGE, linewidth=2, label=f"Median age = {df['Age'].median():.0f}")
plt.title("Age distribution")
plt.xlabel("Age (years)")
plt.ylabel("Number of players")
plt.legend()
plt.tight_layout()
plt.show()

# 5. capped vs uncapped by role
ct = pd.crosstab(df["Specialism"], df["Cap_Status"])[["Capped", "Uncapped"]]
ax = ct.plot(kind="bar", figsize=(8, 4.5), color=[BLUE, ORANGE])
for c in ax.containers:
    ax.bar_label(c)
plt.title("Capped vs uncapped players by role")
plt.ylabel("Number of players")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 6. correlation heatmap
corr = df[["Age", "Test_Caps", "ODI_Caps", "T20_Caps", "IPL_Matches",
           "Prev_Teams_Count", "Base_Price_Lakh"]].corr()
plt.figure(figsize=(8, 6))
plt.imshow(corr, cmap="Blues", vmin=0, vmax=1)
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=40, ha="right")
plt.yticks(range(len(corr)), corr.columns)
for i in range(len(corr)):
    for j in range(len(corr)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center",
                 color="white" if corr.iloc[i, j] > 0.6 else "black", fontsize=9)
plt.title("Correlation between numeric columns")
plt.tight_layout()
plt.show()
```

## 4. Five Insights

Every insight is calculated from the data, so the numbers always match the dataset.

1. **Nationality** - how much of the pool is Indian?
2. **Roles** - which role is the scarcest?
3. **Pricing** - how are base prices spread?
4. **Age** - capped vs uncapped players
5. **Experience and price** - do experienced players start at higher prices? (correlation shows two things move together, it does not prove cause)

```python
N = len(df)

# 1. Nationality
indian = (df["Nationality"] == "Indian").sum()
overseas = df.loc[df["Country"] != "India", "Country"].value_counts()
print(f"1. India dominates the pool: {indian} of {N} players ({indian / N:.0%}) are Indian. "
      f"The biggest overseas groups are {overseas.index[0]} ({overseas.iloc[0]}) and "
      f"{overseas.index[1]} ({overseas.iloc[1]}), from {df['Country'].nunique()} countries in total.\n")

# 2. Roles
roles = df["Specialism"].value_counts()
print(f"2. Role mix: {roles.index[0]}s ({roles.iloc[0]}, {roles.iloc[0] / N:.0%}) and "
      f"{roles.index[1]}s ({roles.iloc[1]}, {roles.iloc[1] / N:.0%}) make up most of the list, "
      f"while {roles.index[-1]}s are the rarest with only {roles.iloc[-1]} players ({roles.iloc[-1] / N:.0%}).\n")

# 3. Pricing
top_price = df["Base_Price_Lakh"].max()
top = df[df["Base_Price_Lakh"] == top_price]
at_30 = (df["Base_Price_Lakh"] == 30).sum()
unc_at_30 = (df.loc[df["Cap_Status"] == "Uncapped", "Base_Price_Lakh"] == 30).mean()
print(f"3. Pricing is bottom-heavy: {at_30} players ({at_30 / N:.0%}) have the Rs 30 lakh minimum "
      f"base price and {unc_at_30:.0%} of uncapped players start there. Only {len(top)} players "
      f"({len(top) / N:.0%}) are in the top Rs {top_price / 100:.0f} crore tier; "
      f"{(top['Cap_Status'] == 'Capped').mean():.0%} of them are capped and "
      f"{(top['Nationality'] == 'Overseas').mean():.0%} are overseas.\n")

# 4. Age
age_by_cap = df.groupby("Cap_Status")["Age"].median()
youngest = df.loc[df["Age"].idxmin()]
oldest = df.loc[df["Age"].idxmax()]
print(f"4. Age: the median age is {df['Age'].median():.0f}. Capped players are {age_by_cap['Capped']:.0f} "
      f"on median and uncapped players are {age_by_cap['Uncapped']:.0f}; "
      f"{(df['Age'] <= 25).mean():.0%} of the pool is 25 or younger. "
      f"Youngest: {youngest['Player']} ({youngest['Age']}), oldest: {oldest['Player']} ({oldest['Age']}).\n")

# 5. IPL experience and price
no_ipl = (df["IPL_Matches"] == 0).mean()
r_ipl = df["IPL_Matches"].corr(df["Base_Price_Lakh"])
r_caps = df["Total_Intl_Caps"].corr(df["Base_Price_Lakh"])
team_counts = df["Prev_IPL_Teams"].str.split(",").explode()
team_counts = team_counts[team_counts != ""].value_counts()
most_exp = df.loc[df["IPL_Matches"].idxmax()]
print(f"5. Experience and price move together: {no_ipl:.0%} of players have never played an IPL match; "
      f"base price correlates with IPL matches (r = {r_ipl:.2f}) and international caps (r = {r_caps:.2f}). "
      f"{team_counts.index[0]} has the most former players in the pool ({team_counts.iloc[0]}), and the most "
      f"experienced player is {most_exp['Player']} with {most_exp['IPL_Matches']} IPL matches.")
```
